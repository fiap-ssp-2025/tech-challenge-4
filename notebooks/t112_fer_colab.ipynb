## 2. Inspeção do dataset (T104)

```python
print('rows', len(df))
print(df['label'].value_counts())
print(df['split'].value_counts())
print('actors', df['actor'].nunique())
# Zero ator cruzando splits
leak = df.groupby('actor')['split'].nunique()
assert (leak == 1).all(), leak[leak > 1]
minority = df['label'].value_counts(normalize=True).min()
print(f'minority={minority:.3f}')
assert minority >= 0.40
```

```python
def show_samples(frame, n=8):
    sample = frame.sample(min(n, len(frame)), random_state=0)
    fig, axes = plt.subplots(2, 4, figsize=(10, 5))
    for ax, (_, row) in zip(axes.flat, sample.iterrows()):
        path = Path(row['path'])
        if not path.is_file():
            path = FACES_DIR / Path(row['path']).name
        ax.imshow(Image.open(path))
        ax.set_title(f"{row['label']}\n{row['emotion']}")
        ax.axis('off')
    plt.tight_layout()

show_samples(df)
```
